## Task 1 - Text to Emoji (1 point)

Write a Python program that takes input text containing emotions and converts them into corresponding emojis. Your program should replace specific text representations of emotions with appropriate emoji symbols. (Do mapping of all emojis, also you can use demoji library) Your code should work for any text input



  ```
      **Sample Input Text:**
      # My heart swells with happiness as I bask in the warmth of your presence, grateful for every moment shared together. Yet, a twinge of sadness tugs at my soul. Despite the challenges, I'm excited for the future, determined to overcome any obstacles that come our way.


      **Output Example:**
      My ❤️ swells with happiness as I bask in the warmth of your presence, grateful for every moment shared together. Yet, a twinge of 😞 tugs at my soul. Despite the challenges, I'm 😃 for the future, determined to overcome any obstacles that come our way.
  ```

````python
def convert_to_emojis(text):
    emoji_map = {
        "happy": "😊",
        "happiness": "😊",
        "sad": "😢",
        "sadness": "😞",
        "love": "❤️",
        "heart": "❤️",
        "excited": "😃",
        "angry": "😡",
        "angrily": "😠",
        "fear": "😨",
        "scared": "😱",
        "crying": "😭",
        "surprised": "😲",
        "grateful": "🙏",
        "determined": "💪",
        "laughing": "😂",
        "joy": "😁",
        "confused": "😕",
        "bored": "😐",
        "tired": "😴",
        "sick": "🤒",
        "cool": "😎",
        "kiss": "😘",
        "wink": "😉",
        "angry": "😠",
        "nervous": "😬",
        "blessed": "😇"
    }

    words = text.split()
    new_words = []
    for word in words:
        key = word.lower().strip(".,!?")  # Normalize word
        if key in emoji_map:
            # Replace while preserving punctuation
            emoji = emoji_map[key]
            punct = ''.join(c for c in word if not c.isalnum())
            new_words.append(emoji + punct)
        else:
            new_words.append(word)
    return ' '.join(new_words)

# Sample Input
input_text = ("My heart swells with happiness as I bask in the warmth of your presence, "
              "grateful for every moment shared together. Yet, a twinge of sadness tugs at my soul. "
              "Despite the challenges, I'm excited for the future, determined to overcome any obstacles that come our way.")

# Output
output_text = convert_to_emojis(input_text)
print(output_text)
````

### 1: Concatenate all three files into a single .txt file 

````python
import pandas as pd

users_df = pd.read_csv("datasets/users.txt")
ratings_df = pd.read_csv("datasets/ratings.txt")
books_df = pd.read_csv("datasets/books.txt")

user_rat_df = pd.merge(users_df,ratings_df, how='outer', on="user_id")
df = pd.merge(user_rat_df,books_df, how='outer', on="ISBN")

df
````

### 2: Book with the highest and lowest number of reviews

````python
review_counts = df['title'].value_counts()
most_reviewed = review_counts.idxmax()
least_reviewed = review_counts.idxmin()
print("Book with the highest number of reviews:", most_reviewed)
print("Book with the lowest number of reviews:", least_reviewed)
````

### 3: Find out which book has the highest average rating - but include only books that have at least 100 reviews!

````python
rating_stats = df.groupby('title').agg({'rating': ['mean', 'count']})
rating_stats.columns = ['average_rating', 'review_count']
qualified_books = rating_stats[rating_stats['review_count'] >= 100]
top_rated_book = qualified_books['average_rating'].idxmax()
print("\nHighest average rated book (min 100 reviews):", top_rated_book)
````

### 4: Highest rated book per country

````python
country_rating_stats = df.groupby(['country', 'title']).agg({'rating': 'mean'}).reset_index()

top_books_by_country = country_rating_stats.loc[country_rating_stats.groupby('country')['rating'].idxmax()]
print("\nHighest rated book in each country:")
pd.DataFrame(top_books_by_country)
````

## Task 3 - Automating University Admission Process (1.5 point)

Develop a Python program using OOP to automate the admission process for a university based on the marks scored in a qualifying exam. Each student is represented by a unique student ID, along with their age and marks obtained in the exam.

**Data Validity Criteria:**

1. **Age:** The age of the student must be greater than 17 years.
2. **Marks:** The marks obtained by the student must be within the range of 0 to 100 (both inclusive).

**Admission Qualification Criteria:**

A student qualifies for admission if the following conditions are met:

1. The age and marks of the student are valid according to the criteria mentioned above.
2. The marks obtained by the student are 35 or more.

**Scholarship Calculation:**

Based on the marks obtained by the student, the following scholarship percentages are applicable:

*   96-100: 100% Scholarship
*   86-95: 75% Scholarship
*   76-85: 50% Scholarship
*   60-75: 20% Scholarship
*   50-60: 10% Scholarship
*   Below 50: No scholarship; full tuition fees apply.


Write a python program to represent the students seeking admission in the university with atleast 5 students input test

````python
class Student:
    def __init__(self, student_id, age, marks):
        self.student_id = student_id
        self.age = age
        self.marks = marks
        self.is_qualified = False
        self.scholarship_percentage = 0
        self.reason = ""
        self.check_qualification()

    def validate_age(self):
        return self.age > 17

    def validate_marks(self):
        return 0 <= self.marks <= 100

    def check_qualification(self):
        if not self.validate_age():
            self.reason = "Age must be greater than 17"
            return
        if not self.validate_marks():
            self.reason = "Marks must be between 0 and 100"
            return
        if self.marks < 35:
            self.reason = "Marks less than 35"
            return
        self.is_qualified = True
        self.set_scholarship()

    def set_scholarship(self):
        if self.marks >= 96:
            self.scholarship_percentage = 100
        elif self.marks >= 86:
            self.scholarship_percentage = 75
        elif self.marks >= 76:
            self.scholarship_percentage = 50
        elif self.marks >= 60:
            self.scholarship_percentage = 20
        elif self.marks >= 50:
            self.scholarship_percentage = 10
        else:
            self.scholarship_percentage = 0

    def display_student(self):
        print(f"Student ID: {self.student_id}")
        print(f"Age: {self.age}")
        print(f"Marks: {self.marks}")
        if self.is_qualified:
            print("Qualified for Admission: Yes")
            print(f"Scholarship Percentage: {self.scholarship_percentage}%")
        else:
            print("Qualified for Admission: No")
            print(f"Reason: {self.reason}")
        print()


students = [
    Student(1, 18, 97),
    Student(2, 18, 90),
    Student(3, 17, 90),
    Student(4, 18, 101),
    Student(5, 18, 30),
    Student(6, 19, 85),
    Student(7, 20, 65),
    Student(8, 18, 55),
    Student(9, 19, 45)
]

for student in students:
    student.display_student()
````

## Task 4 - TollBooth (1.5 point)

Imagine a tollbooth at a bridge. Vehicle passing by the booth are expected to pay or freely pass with a toll fee based on their vehicle type:


*   **Bikes, Two-Wheeler and Scooter** pass for **free** (pay no toll).
*   **Cars** are charged **500 rubles**.
*   **Trucks/Buses** are charged **1200 rubles**.
*   **Heavy-duty equipment vehicles** are charged **1500 rubles**.


The tollbooth keeps track of the number of vehicles that have passed and the total amount of money collected.

Model this tollbooth with a class called ***TollBooth***. The two data items are an unsigned integer to hold the total number of vehicles and an integer to hold the total amount of money collected. A constructor initializes both of these to zero.

Implement the following member functions:

1. **payVehicle(int toll)**: Increments the vehicle total and adds the specified toll amount to the cash total.
2. **noPayVehicle()**: Increments the vehicle total but adds nothing to the cash total.
3. **display()**: Displays the total number of vehicles and the total amount of money collected.

Create an object of this class in the main function. Continuously display a menu to the user, allowing them to choose the type of vehicle passing by. Update the tollbooth accordingly and display the total statistics. Keep displaying the menu until the user chooses to exit the program.

````python
class TollBooth:
    def __init__(self):
        self.total_vehicles = 0
        self.total_cash = 0

    def payVehicle(self, toll):
        self.total_vehicles += 1
        self.total_cash += toll

    def noPayVehicle(self):
        self.total_vehicles += 1

    def display(self):
        print(f"Total Vehicles: {self.total_vehicles}")
        print(f"Total Cash Collected: {self.total_cash} rubles")

tollbooth = TollBooth()
while True:
    print("\nTollBooth Menu:")
    print("1. Bike/Two-Wheeler/Scooter (Free)")
    print("2. Car (500 rubles)")
    print("3. Truck/Bus (1200 rubles)")
    print("4. Heavy-duty equipment (1500 rubles)")
    print("5. Exit")
    choice = input("Enter choice (1-5): ")
    
    if choice == '1':
        tollbooth.noPayVehicle()
        tollbooth.display()
    elif choice == '2':
        tollbooth.payVehicle(500)
        tollbooth.display()
    elif choice == '3':
        tollbooth.payVehicle(1200)
        tollbooth.display()
    elif choice == '4':
        tollbooth.payVehicle(1500)
        tollbooth.display()
    elif choice == '5':
        print("Final Statistics:")
        tollbooth.display()
        break
    else:
        print("Invalid choice, please try again.")
````
